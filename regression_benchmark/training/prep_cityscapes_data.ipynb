```python
import os
import json
from tqdm import tqdm

CLASS_MAP = {
    "person": 0,
    "rider": 1,
    "car": 2,
    "truck": 3,
    "bus": 4,
    "train": 5,
    "motorcycle": 6,
    "bicycle": 7,
}
```

```python
# Train

# paths
json_dir = "/path/to/datasets/Cityscapes/train/json/"
label_dir = "/path/to/datasets/Cityscapes/train/labels/"

os.makedirs(label_dir, exist_ok=True)


for file in tqdm(os.listdir(json_dir)):
    if not file.endswith(".json"):
        continue

    with open(os.path.join(json_dir, file)) as f:
        data = json.load(f)

    H = data["imgHeight"]
    W = data["imgWidth"]

    yolo_lines = []

    for obj in data["objects"]:
        label = obj["label"]

        # skip unknown classes
        if label not in CLASS_MAP:
            continue

        x1, y1, x2, y2 = obj["bbox"]

        # convert
        xc = (x1 + x2) / 2 / W
        yc = (y1 + y2) / 2 / H
        w  = (x2 - x1) / W
        h  = (y2 - y1) / H

        # filter invalid boxes (optional but recommended)
        if w <= 0 or h <= 0:
            continue

        cls_id = CLASS_MAP[label]

        yolo_lines.append(f"{cls_id} {xc:.6f} {yc:.6f} {w:.6f} {h:.6f}")

    # save
    out_file = os.path.join(label_dir, file.replace(".json", ".txt"))
    with open(out_file, "w") as f:
        f.write("\n".join(yolo_lines))
```
